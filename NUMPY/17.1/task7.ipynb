```python
# Task 7: Customized Charts
# Create a bar chart with customized colors, labels, and grid lines.

import matplotlib.pyplot as plt

products = ["latop","mobile","tablet","headphone","watch"]
sales = [50, 80, 40, 70, 30]

colors = ["red","blue","green","orange","purple"]

plt.bar(
    products,
    sales,
    color=colors,
    edgecolor="black"
)

plt.xlabel("products")
plt.ylabel("sales")
plt.title("sales of five products")


plt.show()
```
