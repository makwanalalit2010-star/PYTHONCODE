```python
# Task 6: Combined Chart
# Plot both bar and line chart together showing revenue and profit for 6 months.
import matplotlib.pyplot as plt

months = ["January", "February", "March", "April", "May", "June"]

revenue = [100, 120, 150, 130, 170, 200]
profit = [20, 25, 35, 30, 40, 50]

plt.bar(months, revenue, label="Revenue")
plt.plot(months, profit, marker="o", label="Profit")

plt.xlabel("Months")
plt.ylabel("Amount")
plt.title("Monthly Revenue and Profit")

plt.legend()
plt.grid(False)

plt.show()
```
