```python
# Task 2: Create a Line Chart
# Plot monthly temperature changes in a city for 12 months.

import matplotlib.pyplot as plt

moths = [
    "january",
    "february",
    "march",
    "april",
    "may",
    "june",
    "july",
    "august",
    "september",
    "october",
    "november",
    "december"
]

temperatures = [18, 19, 21, 23, 26, 28, 30, 29, 27, 24, 21, 19]

plt.plot(moths, temperatures)

plt.title("Monthly Temperature Changes in a City")
plt.xlabel("Months")
plt.ylabel("Temperature (°C)")

plt.show()
plt.grid(True)
```
