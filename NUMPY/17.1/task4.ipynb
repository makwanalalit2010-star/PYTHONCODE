```python
# Task 4: Create a Pie Chart
# Show a pie chart representing the percentage distribution of different fruits sold in a store.

import matplotlib.pyplot as plt

fruits = ["apple","banana","mango","orange","grapes"]
sales = [20,15,20,10,35]

plt.pie(
    sales,s
    labels=fruits,
    autopct='%1.1f%%'
    )

plt.title("Percentage Distribution of Fruits Sold")
plt.show()

```
