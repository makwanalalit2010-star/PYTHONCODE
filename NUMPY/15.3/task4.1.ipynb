```python
# Task 4.1: Broadcasting a Scalar to an Array
# Create a NumPy array prices with the following values: [10, 20, 30].
# Increase all prices by 5 using broadcasting.

import numpy as np

prices = np.array([10,20,30])

new_price = prices + 5

print("original prices:")
print(prices)

print("\nnew prices:")
print(new_price)
```
