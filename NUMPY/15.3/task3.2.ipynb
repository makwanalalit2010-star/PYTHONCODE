```python
# Task 3.2: Reshaping a 2D array to 1D (Flattening)
# Using the numbers_2d array from the previous task, flatten it back into a 1D array.

import numpy as np

numbers_1d = np.arange(1, 13)

number_2d = numbers_1d.reshape(3, 4)

print("2d array:")
print(number_2d)

numbers_flat = number_2d.flatten()
print("\n flattend 1d array:")
print(numbers_flat)
```
