```python
# Task 2.2: Masking for Modification
# Using the temperatures array again, use the is_hot mask to change all temperatures above 30 degrees Celsius to 31 degrees Celsius.

import numpy as np

tem = np.array([28, 32, 25, 35, 30, 22, 38])

is_hot = tem > 30

tem[is_hot] = 31

print("updated tem:")
print(tem)
```
