```python
# Task 2: Aggregation and Grouping
# 2.1 Group the dataset by a categorical column (e.g., Gender, Class).
# 2.2 Calculate aggregate statistics (mean, sum, count) for each group.
# 2.3 Display the group with the highest average value of a selected column.

# 2.1 Group the dataset by a categorical column (e.g., Gender, Class).

import pandas as pd

# 2.1 Load dataset
df = pd.read_csv('titanic.csv')

# Group dataset by Gender
grouped = df.groupby('Sex')

# 2.2 Calculate aggregate statistics
summary = grouped.agg({
    'Age': ['mean', 'count'],
    'Fare': ['mean', 'sum']
})

print("Group-wise Summary:")
print(summary)

# 2.3 Find group with highest average Fare
highest_avg_fare_group = df.groupby('Sex')['Fare'].mean().idxmax()
highest_avg_fare = df.groupby('Sex')['Fare'].mean().max()

print("\nGroup with highest average Fare:")
print(highest_avg_fare_group)
print("Average Fare:", highest_avg_fare)
```
