```python
class SalesDataAnalyzer:
    def __init__(self, filepath=None):
        """Initializes analyzer instance with data, figure, and filepath attributes.
        Arguments: filepath (optional string path to CSV file)
        Return value: None (initializes attributes)"""
        self.data = None
        self.currentfig = None
        self.filepath = filepath
        self.dummy_df = None
        self.dummy_file = None

    def __str__(self):
        """Returns string representation of the analyzer showing loaded record count.
        Arguments: None
        Return value: String message with loaded records count or no data loaded status"""
        if self.data is not None:
            return f"SalesDataAnalyzer:{len(self.data)} records loaded"
        return "SalesDataAnalyzer: No dataset loaded."

    def __del__(self):
        """Destructor method that prints cleanup message when session terminates.
        Arguments: None
        Return value: None (prints cleanup message)"""
        print("SalesDataAnalyzer session closed. Resources cleaned up")

    def load_data(self, filepath=None):
        """Loads sales dataset from CSV file into pandas DataFrame.
        Arguments: filepath (optional string path to CSV file)
        Return value: None (stores loaded DataFrame in self.data)"""
        if filepath is None:
            filepath = input("Enter the path of dataset (CSV file): ")
        try:
            self.data = pd.read_csv(filepath.strip())
            self.filepath = filepath
            print(self.data)
            print("""
            
            Data Loaded Sucessfully!!!""")
            print(self)
        except FileNotFoundError:
            print("File Does Not Exist,Please Check The Entered Path")
        except Exception as e:
            print("Error Detected In Selected File")
            
    def explore_data(self):
        """Displays dataset structure including head, tail, columns, dtypes, and info.
        Arguments: None
        Return value: None (prints exploratory dataset details)"""
        if self.data is None:
            print("Error:file not loaded,is Empty or File is Invalid,please Check...")
        else:
            print("""
            ---Explore Dataset---
            1. Display first 5 rows
            2. Display last 5 rows
            3. Display column names
            4. Display data types
            5. Display basic info""")
            usrinput = input("Enter your choice: ")
            if usrinput == "1":
                print(self.data.head())
            elif usrinput == "2":
                print(self.data.tail())
            elif usrinput == "3":
                print(self.data.columns.tolist())
            elif usrinput == "4":
                print(self.data.dtypes)
            elif usrinput == "5":
                self.data.info()
            else:
                print("invaled choice selected!")

    def perform_dataframe_operations(self):
        """Provides sub-menu to perform mathematical operations, combine, split, or search/sort/filter.
        Arguments: None
        Return value: None (calls selected DataFrame operation method)"""
        if self.data is None:
            print("Error:file not loaded,is Empty or File is Invalid,please Check...")
            return
        print("""
        ---Perform DataFrame Operations---
        1. Mathematical Operations (NumPy & Slicing)
        2. Combine Data
        3. Split Data
        4. Search, Sort & Filter
        """)
        op_choice = input("Enter your choice: ")
        if op_choice == "1":
            self.mathematical_operations()
        elif op_choice == "2":
            self.combine_data()
        elif op_choice == "3":
            self.split_data()
        elif op_choice == "4":
            self.search_sort_filter()
        else:
            print("invaled choice selected!")

    def clean_data(self):
        """Handles missing values by viewing, filling with mean, dropping, or replacing.
        Arguments: None
        Return value: None (updates self.data and prints cleaned DataFrame)"""
        if self.data is None:
            print("Error:file not loaded,is Empty or File is Invalid,please Check...")
        else:
            print("""
            ---Handle Missing Data---
            1.Display rows with missing values
            2.Fill missing values with mean
            3.Drop rows with missing values
            4.Replace missing values with specific value
            """)
            usrinput = input("Enter your choice: ")
            if usrinput == "1":
                if self.data.isna().values.any() == True:
                    print(f"""
                    data containing missing values in database are:
                    {self.data[self.data.isna().any(axis=1)]}""")
                else:
                    print("No Missing Value Found!")
            elif usrinput == "2":
                if self.data.isna().values.any() == True:
                    data_mean = self.data.mean(numeric_only=True)
                    self.data.fillna(data_mean, inplace=True)
                    print("Missing values filled with the mean sucessfully")
                    print(self.data)  
                else:
                    print("No Missing Values Found")
            elif usrinput == "3":
                if self.data.isna().values.any() == True:
                    self.data.dropna(inplace=True)
                    print("Rows With Missing Values Dropped Sucessfully")
                    print(self.data)
                else:
                    print("No Missing Values Found")
            elif usrinput == "4":
                if self.data.isna().values.any() == True:
                    val = float(input("Enter Replacement numeric Value: "))
                    self.data.fillna(val, inplace=True)
                    print("Missing Values Replaced By Given Value Sucessfully")
                    print(self.data)
                else:
                    print("No Missing Values Found")
            else:
                print("invaled choice selected!")

    def mathematical_operations(self):
        """Converts sales to NumPy array, performs slicing, and calculates Tax and Profit Margin.
        Arguments: None
        Return value: None (adds Tax and Profit Margin columns and prints data)"""
        if self.data is None:
            print("Error:file not loaded,is Empty or File is Invalid,please Check...")
        else:
            sales_array = np.array(self.data['Sales'])
            print(f"""
            First Sale Amount
            {sales_array[0]}
            .......
            First Three Sales Amount
            {sales_array[:3]}
            """)
            self.data['Tax'] = self.data['Sales'] * (18 / 100)
            self.data['Profit Margin'] = (self.data['Profit'] / self.data['Sales']) * 100
            print(self.data)

    def combine_data(self, dummy_file=None, dummy_df=None):
        """Combines loaded DataFrame with another dataset using concatenation.
        Arguments: dummy_file (optional string CSV path), dummy_df (optional DataFrame)
        Return value: None (concatenates new data into self.data and prints result)"""
        if dummy_df is not None:
            self.dummy_df = dummy_df
        elif dummy_file is None:
            dummy_file = input("Enter the path of another dataset (CSV file): ")

        if self.dummy_df is None and dummy_file is not None:
            try:
                self.dummy_df = pd.read_csv(dummy_file.strip())
                self.dummy_file = dummy_file
                print(self.dummy_df)
                print("""
                
                Dummy File And Data Loaded Sucessfully!!!""")
            except FileNotFoundError:
                print("File Does Not Exist,Please Check The Entered Path")
            except Exception as e:
                print("Error Detected In Selected File")    

        if self.data is not None and self.dummy_df is not None:
            self.data = pd.concat([self.data, self.dummy_df], ignore_index=True)  
            print(f"""

            Dataframe Concatination Done Sucessfully!!!
            """)
            print(self.data)
        else:
            print("Error:file not loaded,is Empty or File is Invalid,please Check...")

    def split_data(self):
        """Splits DataFrame into subsets based on Region or Product selection.
        Arguments: None
        Return value: None (prints filtered subset and group DataFrames)"""
        if self.data is None:
            print("Error:file not loaded,is Empty or File is Invalid,please Check...")
        else:
            print("""
            On Which basis do you want to split data
            1.Region
            2.Product
            """)
            userinp = input("enter your choice: ")
            if userinp == "1":
                print(f"Regions in the main loaded dataframe: {self.data['Region'].unique()}") 
                reg = input("Enter Region: ")
                region_df = self.data[self.data['Region'] == reg]
                print(region_df)
                print("") 
                groups = dict(tuple(self.data.groupby('Region')))
                for name, group_df in groups.items():
                    print(f"---Region: {name}---")
                    print(group_df)
                    print("") 
            elif userinp == "2":
                print(f"Products in the main loaded dataframe: {self.data['Product'].unique()}")
                prd = input("Enter Product Name:")
                product_df = self.data[self.data['Product'] == prd]
                print(product_df)
                print("")
                groups = dict(tuple(self.data.groupby('Product')))
                for name, group_df in groups.items():
                    print(f"---Product: {name}---")
                    print(group_df)
                    print("")
            else:
                print("invaled choice selected!") 

    def search_sort_filter(self):
        """Searches product names, sorts by Sales/Profit, and filters by sales amount.
        Arguments: None
        Return value: None (prints searched, sorted, or filtered rows)"""
        if self.data is None:
            print("Error:file not loaded,is Empty or File is Invalid,please Check...")
        else:
            print("""
            ---Search,Sort & Filter---
            1. Search By Product Name
            2.Sort Data (by Sales or Profit)
            3.Filter Data (e.g. based on min. sale)""")
            userinp = input("Enter The Choice: ")
            if userinp == "1":
                prd_name = input("Enter the product to search: ")
                product_df = self.data[self.data['Product'] == prd_name]
                print(product_df)
                print("") 
            elif userinp == "2":
                print("""
                On Which Basic Do You Want To Sort The Data:
                1.Sales
                2.Profit""")
                userinp1 = input("Enter Your Choice (Sales,Profit):")
                userinp2 = input("1.Ascending Order , 2.Descinding Order??: ")
                if userinp2 == "1":
                    print(self.data.sort_values(by=userinp1, ascending=True))
                    print("")
                elif userinp2 == "2":
                    print(self.data.sort_values(by=userinp1, ascending=False))
                    print("")
                else:
                    print("Invalid Choice Selected!!")
            elif userinp == "3":
                min_sale = float(input("Enter The Minimum Sales Amount: "))
                print(self.data[self.data['Sales'] >= min_sale])
                print("")
            else:
                print("invaled choice selected!") 

    def generate_descriptive_statistics(self):
        """Provides sub-menu to generate summary statistics, aggregations, or pivot table.
        Arguments: None
        Return value: None (calls selected statistical method)"""
        if self.data is None:
            print("Error:file not loaded,is Empty or File is Invalid,please Check...")
            return
        print("""
        ---Generate Descriptive Statistics---
        1. Statistical Analysis (Summary, Std, Var, Quantiles)
        2. Aggregation Functions & Transform
        3. Create Pivot Table
        """)
        stat_choice = input("Enter your choice: ")
        if stat_choice == "1":
            self.statistical_analysis()
        elif stat_choice == "2":
            self.aggregate_functions()
        elif stat_choice == "3":
            self.create_pivot_table()
        else:
            print("invaled choice selected!")

    def statistical_analysis(self):
        """Calculates descriptive summary, standard deviation, variance, and quantiles.
        Arguments: None
        Return value: None (prints descriptive statistical metrics)"""
        if self.data is None:
            print("Error:file not loaded,is Empty or File is Invalid,please Check...")
        else:
            print("--Full Summary Of Dataframe--")
            print(self.data.describe())
            print("")
            print("--Standard Deviation--")
            print(self.data[['Sales', 'Profit']].std())
            print("")
            print("--Variance--")
            print(self.data[['Sales', 'Profit']].var())
            print("")
            print("--Quantiles (Percentile) (1.25th percentile ,50th percentile(Median) ,75th percentile)--")
            print(self.data[['Sales', 'Profit']].quantile([0.25, 0.50, 0.75]))
            print("")

    def aggregate_functions(self):
        """Calculates total/average sales, orders, group-by metrics, and regional averages.
        Arguments: None
        Return value: None (updates self.data with Region_Avg_Sales and prints results)"""
        if self.data is None:
            print("Error:file not loaded,is Empty or File is Invalid,please Check...")
        else:
            print("--Total Sales--")
            print(self.data['Sales'].sum())
            print("")
            print("--Average Sales--")
            print(self.data['Sales'].mean())
            print("")
            print("--Total Orders--")
            print(self.data['OrderID'].count())
            print("")
            print("--Region Wise Total Sale,Average Sale and No. Of Sales--")
            print(self.data.groupby('Region')['Sales'].agg(['sum', 'mean', 'count']))
            print("")
            print("--Use Of Transform Function(repeat agg fxn) For regional average sales--")
            self.data['Region_Avg_Sales'] = self.data.groupby('Region')['Sales'].transform('mean')
            print(self.data)
            print("")

    def create_pivot_table(self):
        """Creates and displays pivot table between Region and Product with sum or mean.
        Arguments: None
        Return value: None (prints pivot table DataFrame)"""
        if self.data is None:
            print("Error:file not loaded,is Empty or File is Invalid,please Check...")
        else:
            print("Which value do you want to get info about in pivot table of Region and Product")
            userinp1 = input("Enter your Choice (Sales/Profit): ")
            print("""
            Which operation do you want to perform with value in pivot table??
            1.Sum
            2.Mean""")
            userinp2 = input("Enter your Choice: ")
            if userinp2 == "1":
                pivot_df = pd.pivot_table(
                    self.data,
                    values=userinp1,
                    index='Region',
                    columns='Product',
                    aggfunc='sum',
                    fill_value=0
                )
                print(pivot_df)
                print("")
            elif userinp2 == "2":
                pivot_df = pd.pivot_table(
                    self.data,
                    values=userinp1,
                    index='Region',
                    columns='Product',
                    aggfunc='mean',
                    fill_value=0
                )
                print(pivot_df)
                print("")
            else:
                print("Invalid Choice Selected!!")

    def visualize_data(self):
        """Generates line plot, bar plot, scatter plot, histogram, or heatmap charts.
        Arguments: None
        Return value: None (displays plot and stores figure in self.currentfig)"""
        if self.data is None:
            print("Error:file not loaded,is Empty or File is Invalid,please Check...")
        else:
            print("""
            --Data Visualization Options--
            1.Line Plot(Sales Trend Over Time)
            2.Bar Plot(Product/Region Comparison)
            3.Scatter Plot (Sales vs Profit Relationship)
            4.Histogram (Distribution of Sales)
            5.Heatmap (Correlation / Pivot Table)
            6.Pie Chart
            """)
            usrinp1 = input("Enter Your Choice: ")
            if usrinp1 == "1":
                plt.figure(figsize=(10, 5))
                plt.plot(self.data['Date'], self.data['Sales'], marker='o', color='b', linestyle='-')
                plt.xticks(rotation=45)
                plt.title("Sales Trend Over Time")
                plt.xlabel("Date")
                plt.ylabel("Sales")
                self.currentfig = plt.gcf()
                plt.show()
            elif usrinp1 == "2":
                plt.figure(figsize=(10, 5))
                prod_sales = self.data.groupby('Product')['Sales'].sum()
                plt.bar(prod_sales.index, prod_sales.values, color='teal')
                plt.xticks(rotation=45)
                plt.title("Bar Plot (Product/Region Comparison)")
                plt.xlabel("Product")
                plt.ylabel("Total Sales")
                self.currentfig = plt.gcf()
                plt.show()
            elif usrinp1 == "3":
                plt.figure(figsize=(8, 5))
                plt.scatter(self.data['Sales'], self.data['Profit'], marker='o', color='coral', s=60)
                plt.title("Sales vs Profit Relationship")
                plt.xlabel("Sales Amount")
                plt.ylabel("Profit Amount")
                self.currentfig = plt.gcf()
                plt.show()
            elif usrinp1 == "4":
                plt.figure(figsize=(10, 5))
                plt.hist(self.data['Sales'], bins=5, color='skyblue', edgecolor='black', linewidth=2)
                plt.title("Distribution of Sales")
                plt.xlabel("Sales Range")
                plt.ylabel("Number of Transactions (Frequency)")
                self.currentfig = plt.gcf()
                plt.show()
            elif usrinp1 == "5":
                plt.figure(figsize=(8, 6))
                corr_matrix = self.data[['Sales', 'Profit']].corr()
                sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")
                plt.title("Correlation Heatmap (Sales & Profit)")
                self.currentfig = plt.gcf()
                plt.show()
            elif usrinp1 == "6":
                plt.figure(figsize=(7, 7))
                region_sales = self.data.groupby('Region')['Sales'].sum()
                plt.pie(region_sales.values, labels=region_sales.index, autopct='%1.1f%%', startangle=90)
                plt.title("Sales Distribution by Region")
                self.currentfig = plt.gcf()
                plt.show()
            else:
                print("invaled choice selected!")

    def save_visualization(self, filename=None):
        """Saves active Matplotlib/Seaborn plot to an image file.
        Arguments: filename (optional string path, e.g. sales_chart.png)
        Return value: None (saves figure file to disk)"""
        if self.currentfig is None:
            print("No plot has been generated yet to save!")
        else:
            if filename is None:
                filename = input("Enter filename to save plot (eg.XYZ.png): ")
            self.currentfig.savefig(filename.strip(), bbox_inches='tight')
            print(f"Visualization saved successfully as {filename}!")
```

```python
analyzer = SalesDataAnalyzer()
while True:
        print("""
        1. Load Dataset
        2. Explore Data
        3. Perform DataFrame Operations
        4. Handle Missing Data
        5. Generate Descriptive Statistics
        6. Data Visualization
        7. Save Visualization
        8. Exit
        """)
        choice = input("Enter your choice (1-8): ")
        if choice == "1":
            analyzer.load_data()
        elif choice == "2":
            analyzer.explore_data()
        elif choice == "3":
            analyzer.perform_dataframe_operations()
        elif choice == "4":
            analyzer.clean_data()
        elif choice == "5":
            analyzer.generate_descriptive_statistics()
        elif choice == "6":
            analyzer.visualize_data()
        elif choice == "7":
            analyzer.save_visualization()
        elif choice == "8":
            methods = [
                analyzer.__init__,
                analyzer.__str__,
                analyzer.__del__,
                analyzer.load_data,
                analyzer.explore_data,
                analyzer.perform_dataframe_operations,
                analyzer.clean_data,
                analyzer.mathematical_operations,
                analyzer.combine_data,
                analyzer.split_data,
                analyzer.search_sort_filter,
                analyzer.generate_descriptive_statistics,
                analyzer.statistical_analysis,
                analyzer.aggregate_functions,
                analyzer.create_pivot_table,
                analyzer.visualize_data,
                analyzer.save_visualization]
            for m in methods:
                print(f"Function: {m.__name__}()")
                print(m.__doc__)
                    
                print("-" * 35)
            print("Haaaaash Project Doneee,Goodbye!")
            break
        else:
            print("Invalid Choice Selected!!")
```
