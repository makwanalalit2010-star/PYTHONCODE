```python
# 2. Stack multiple 2D arrays using vstack and hstack and then split them back using vsplit and hsplit.
import numpy as np

a = np.array([
    [1,2],
    [3,4]
])

b = np.array([
    [5,6],
    [7,8]
])

v = np.vstack((a, b))

h = np.hstack((a, b))

print("\nvstack:")
print(v)

print("\nhstack:")
print(h)

v_split = np.vsplit(v, 2)

h_split = np.hsplit(h, 2)

print("\nvsplit:")
print(v_split)

print("\nhsplit:")
print(h_split)
```
