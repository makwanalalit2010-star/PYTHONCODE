```python
# 4. Demonstrate the difference between concatenate, stack, and hstack functions using 2D arrays.

import numpy as np

a = np.array([[1,2],
              [3,4]])

b = np.array([[5, 6],
              [7,8]])

print("array A:")
print(a)

print("\narray b:")
print(b)

concat_axis0 = np.concatenate((a, b), axis=0)

print("\nconccatenate (axis=0) :")
print(concat_axis0)

concat_axis1 = np.concatenate((a, b), axis=1)

print("\nconcatenate (axis=1):")
print(concat_axis1)

stack_result = np.stack((a, b))

print("\nstack:")
print(stack_result)

print("stack shape:", stack_result.shape)

hstack_result = np.hstack((a, b))

print("\nhstack:")
print(hstack_result)
```
