```python
#  5. Create a 3D array and flatten it, then reshape it back to its original shape.
import numpy as np

arr = np.arange(1, 25).reshape(2, 3, 4)

print("original 3D arrat:")
print(arr)

flat = arr.flatten()

print("\nflattened array:")
print(flat)

original = flat.reshape(2, 3, 4)

print("\nreshaped array:")
print(original)

print("\n original shope:", arr.shape)
print("flattened shape:", flat.shape)
print("reshaped shape:", original.shape)
```
