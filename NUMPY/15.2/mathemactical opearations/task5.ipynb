```python
# 5. Implement a function using NumPy that calculates the Euclidean distance between two points.

import numpy as np

def euclidean_distance(p1, p2):
    p1 = np.array(p1)
    p2 = np.array(p2)

    distance = np.sqrt(np.sum((p2 - p1) ** 2))

    return distance


p1 = [2, 3]
p2 = [5, 7]

result = euclidean_distance(p1, p2)

print(result)
```
