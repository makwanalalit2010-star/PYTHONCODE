```python
#### 2. Generate a NumPy array with random integers and apply square root, exponential, and logarithmic functions.

import numpy as np

arr = np.random.randint(1,10, size=5)

print("original array:")
print(arr)

print("square root:")
print(np.sqrt(arr))

print("exponential:")
print(np.exp(arr))

print("logarithm:")
print(np.log(arr))
```
