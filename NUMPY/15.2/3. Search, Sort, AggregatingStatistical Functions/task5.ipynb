```python
# 5. Create a NumPy array and count how many elements are positive, negative, or zero.

import numpy as  np

arr = np.array([10,-5,0,20,-8,0,15,-3,-4,-7,-5,0])

print(arr)

positive_count = np.sum(arr > 0)

negative_count = np.sum(arr < 0)

zero_count = np.sum(arr == 0)

print("\n positive number:", positive_count)
print("negative number:", negative_count)
print("zero elements:", zero_count)
```
