```python
# 1. Create a NumPy array and find the index of the maximum and minimum elements.

import numpy as np

arr = np.array([10,25,5,40,13,30])

print(arr)

max_index = np.argmax(arr)

min_index = np.argmin(arr)

print("\nindex of maximum element:", max_index)
print("\nindex of minimum element:", min_index)

print("maximum value:", arr[max_index])
print("minimum value:", arr[min_index])
```
