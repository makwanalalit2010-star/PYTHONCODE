```python
# 3. Generate a random NumPy array and find the mean, median, variance, and standard deviation.

import numpy as np

arr = np.random.randint(1, 101, size=20)

print(arr)

mean = np.mean(arr)

median = np.median(arr)

variance = np.var(arr)

std = np.std(arr)

print("\nmean:", mean)
print("median:", median)
print("variance:", variance)
print("standard deviation:", std)
```
