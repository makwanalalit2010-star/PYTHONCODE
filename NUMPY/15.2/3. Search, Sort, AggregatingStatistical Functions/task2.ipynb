```python
# 2. Sort a 2D array along different axes and display the result.

import numpy as np

arr = np.array([[30,10,20],
                [60,40,50],
                [90,70,80]])

print("original array:")
print(arr)

sort_axis0 = np.sort(arr, axis=0)

print(sort_axis0)

sort_axis1 = np.sort(arr, axis=1)

print(sort_axis1)
```
