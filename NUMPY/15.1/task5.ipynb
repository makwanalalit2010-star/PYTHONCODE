```python
# write a program to demostrate the difference between a python list and a numpy array by perfoming element wise addition on both

import numpy as np

list1 = [1,2,3,4,5]
list2 = [6,7,8,9,10]

arr1 = np.array([1,2,3,4,5])
arr2 = np.array([6,7,8,9,10])


list_sum = list1 + list2
print("sum of two lists: (list1 + list2):")
print(list_sum)  # This will concatenate the two lists instead of performing element-wise addition

arr_sum = arr1 + arr2
print("\n numpy array addition (arr1 + arr2):")
print(arr_sum)  # This will perform element-wise addition
```
