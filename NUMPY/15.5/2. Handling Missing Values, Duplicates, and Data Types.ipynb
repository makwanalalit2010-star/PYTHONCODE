```python
# # 2. Handling Missing Values, Duplicates, and Data Types
# Task 5: Load a DataFrame with missing values and fill them with appropriate values
# Task 6: Drop rows where any value is missing
# Task 7: Identify and remove duplicate rows
# Task 8: Convert a column from string to numeric (handle errors gracefully)
# Task 9: Check and change data types of columns in a DataFrame

# Task 5: Load a DataFrame with missing values and fill them with appropriate values
import pandas as pd

df = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie'],
    'age': [20, None, 19],
    'salary': [50000, 60000, None]
})

df['age'] = df['age'].fillna(df['age'].mean())
df['salary'] = df['salary'].fillna(df['salary'].mean())

print(df)

# Task 6: Drop rows where any value is missing
df = df.dropna()
print(df)

# Task 7: Identify and remove duplicate rows
print(df.duplicated())

df = df.drop_duplicates()

print(df)

# Task 8: Convert a column from string to numeric (handle errors gracefully)

df['salary'] = pd.to_numeric(df['salary'], errors='coerce')

print(df)

# Task 9: Check and change data types of columns in a DataFrame

print(df.dtypes)

df['age'] = df['age'].astype(int)

print(df.dtypes)

```
