```python
# 3. Handling Dates and Times using Pandas
# Task 10: Load a DataFrame containing a 'date' column with string format, and convert it to datetime
# Task 11: Extract year, month, and day from the datetime column
# Task 12: Filter the DataFrame for rows within a specific date range
# Task 13: Create a new column with the difference in days between two datetime columns

# Task 10: Load a DataFrame containing a 'date' column with string format, and convert it to datetime
import pandas as pd

df = pd.DataFrame({
    'date': ['2026-01-15', '2026-05-20', '2026-09-10']
})

df['date'] = pd.to_datetime(df['date'])

print(df)
print(df.dtypes)

# Task 11: Extract year, month, and day from the datetime column
df['year'] = df['date'].dt.year
df['month'] = df['date'].dt.month
df['day'] = df['date'].dt.day

print(df)

# Task 12: Filter the DataFrame for rows within a specific date range
start_date = '2026-01-01'
end_date = '2026-06-30'

filtered_df = df[
    (df['date'] >= start_date) &
    (df['date'] <= end_date)
]

print(filtered_df)

# Task 13: Create a new column with the difference in days between two datetime columns
df = pd.DataFrame({
    'start_date': ['2026-01-10', '2026-02-15', '2026-03-20'],
    'end_date': ['2026-01-15', '2026-02-25', '2026-03-30']
})

df['start_date'] = pd.to_datetime(df['start_date'])
df['end_date'] = pd.to_datetime(df['end_date'])

df['difference_days'] = (
    df['end_date'] - df['start_date']
).dt.days

print(df)
```
