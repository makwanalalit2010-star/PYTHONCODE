```python
import numpy as np


class ArrayProject:

    def __init__(self):
        self.matrix = None

    def create(self):

        print("\n----- ARRAY CREATION -----")
        print("1. One Dimensional")
        print("2. Two Dimensional")
        print("3. Three Dimensional")

        option = int(input("Select option : "))

        data = list(map(int, input("Enter numbers with space : ").split()))

        if option == 1:

            self.matrix = np.array(data)

            print("\n1. Indexing")
            print("2. Slicing")

            action = int(input("Select operation : "))

            if action == 1:
                pos = int(input("Enter index : "))
                print("Result :", self.matrix[pos])

            elif action == 2:
                start = int(input("Enter starting index : "))
                stop = int(input("Enter ending index : "))
                print("Result :", self.matrix[start:stop])

        elif option == 2:

            r = int(input("Enter rows : "))
            c = int(input("Enter columns : "))

            self.matrix = np.array(data).reshape(r, c)

            print("\nCreated Array:")
            print(self.matrix)

            print("\n1. Indexing")
            print("2. Slicing")

            action = int(input("Select operation : "))

            if action == 1:
                pos = int(input("Enter row index : "))
                print("Result :")
                print(self.matrix[pos])

            elif action == 2:
                start = int(input("Enter starting row : "))
                stop = int(input("Enter ending row : "))
                print("Result :")
                print(self.matrix[start:stop])

        elif option == 3:

            depth = int(input("Enter number of arrays : "))
            r = int(input("Enter rows : "))
            c = int(input("Enter columns : "))

            self.matrix = np.array(data).reshape(depth, r, c)

            print("\nCreated Array:")
            print(self.matrix)


    def arithmetic(self):

        print("\n----- MATHEMATICAL OPERATIONS -----")
        print("1. Addition")
        print("2. Subtraction")
        print("3. Multiplication")
        print("4. Division")

        operation = int(input("Choose operation : "))

        data2 = list(
            map(int, input("Enter second array values : ").split())
        )

        another = np.array(data2).reshape(self.matrix.shape)

        operations = {
            1: np.add,
            2: np.subtract,
            3: np.multiply,
            4: np.divide
        }

        if operation in operations:
            answer = operations[operation](self.matrix, another)
            print("\nResult:")
            print(answer)
        else:
            print("Invalid operation")


    def array_operations(self):

        print("\n----- ARRAY OPERATIONS -----")
        print("1. Split Array")
        print("2. Combine Arrays")

        option = int(input("Enter choice : "))

        if option == 1:

            pieces = int(input("How many parts : "))

            output = np.array_split(self.matrix, pieces)

            print("\nSplit Result:")
            for item in output:
                print(item)

        elif option == 2:

            values = list(
                map(int, input("Enter second array : ").split())
            )

            another = np.array(values).reshape(self.matrix.shape)

            combined = np.concatenate((self.matrix, another))

            print("\nCombined Result:")
            print(combined)


    def find_data(self):

        print("\n----- SEARCH / SORT / FILTER -----")
        print("1. Search")
        print("2. Sort")
        print("3. Filter")

        option = int(input("Enter choice : "))

        if option == 1:

            value = int(input("Enter value to search : "))

            location = np.where(self.matrix == value)

            print("Found at :", location)

        elif option == 2:

            print("Sorted Array:")
            print(np.sort(self.matrix))

        elif option == 3:

            value = int(input("Enter minimum value : "))

            filtered = self.matrix[self.matrix > value]

            print("Filtered Values:")
            print(filtered)


    def calculate_statistics(self):

        print("\n----- STATISTICS -----")
        print("1. Total")
        print("2. Average")
        print("3. Smallest")
        print("4. Largest")

        option = int(input("Enter choice : "))

        if option == 1:

            result = np.sum(self.matrix)
            print("Total is:", result)

        elif option == 2:

            result = np.mean(self.matrix)
            print("Average is:", result)

        elif option == 3:

            result = np.min(self.matrix)
            print("Smallest is:", result)

        elif option == 4:

            result = np.max(self.matrix)
            print("Largest is:", result)


# ---------------- MAIN PROGRAM ----------------

project = ArrayProject()

while True:

    print("\n==============================")
    print("       DATA ANALYTICS")
    print("==============================")

    print("1. Create Array")
    print("2. Mathematical Operations")
    print("3. Split / Combine")
    print("4. Search / Sort / Filter")
    print("5. Statistics")
    print("6. Exit")

    menu = input("Enter your choice : ")

    if menu == "1":
        project.create()

    elif menu == "2":
        project.arithmetic()

    elif menu == "3":
        project.array_operations()

    elif menu == "4":
        project.find_data()

    elif menu == "5":
        project.calculate_statistics()

    elif menu == "6":
        print("Thank you for using Data Analytics!")
        break

    else:
        print("Please enter a valid choice.")
```
